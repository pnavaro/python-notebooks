## **While Loop**

- **Syntax**: Don't forget the **`:`** after the condition.
- **Indentation**: The body of the loop **must be indented**.

**Example: Fibonacci Series**
The Fibonacci series is a sequence where each number is the sum of the two preceding ones.

````python
# Fibonacci series: the sum of two elements defines the next
a, b = 0, 1
while b < 500:
    a, b = b, a + b
    print(round(b / a, 5), end=",")
````

## **`if` Statements**

Python uses the following **logical operators** and **comparison operators**:
- **Logical**: `True`, `False`, `and`, `or`, `not`
- **Comparison**: `==`, `is`, `!=`, `is not`, `>`, `>=`, `<`, `<=`

**Note**: Python does **not** have `switch` or `case` statements. Use `if-elif-else` instead.

**Example:**

````python
x = 42
if x < 0:
    x = 0
    print('Negative changed to zero')
elif x == 0:
    print('Zero')
elif x == 1:
    print('Single')
else:
    print('More')
````

## **Exercise: Collatz Conjecture**

**Description**:
The Collatz conjecture is a mathematical sequence defined as follows:
- If the number is **even**, divide it by two.
- If the number is **odd**, triple it and add one.

The conjecture states that no matter what initial positive integer you start with, the sequence will always reach **1**.

**Task**:
- Test the Collatz conjecture for `n = 100000`.
- Count how many steps are needed to reach **1**.

## **Looping Over Iterable Objects**

Python uses the `for` statement to loop over **iterable objects** (e.g., strings, lists, tuples).
If used with a string, it loops over its **characters**.

**Example:**

````python
words = ["cat", "window", "defenestrate"]
for word in words:
    print(word, len(word))
````

**Output:**
```
cat 3
window 6
defenestrate 12
```

## **Exercise: Narcissistic Numbers**

A **narcissistic number** (or Armstrong number) is a number that is equal to the sum of its own digits, each raised to the power of the number of digits.

**Examples**:
- $153 = 1^3 + 5^3 + 3^3$
- $4150 = 4^5 + 1^5 + 5^5 + 0^5$

**Task**:
Find all **3-digit narcissistic numbers**.

## **Exercise: Happy Numbers**

A **happy number** is defined by the following process:
1. Start with a number `n = n₀`.
2. Define a sequence where each subsequent number is the **sum of the squares of the digits** of the previous number.
3. If the sequence reaches **1**, the number is happy.

**Example**:
For `n = 19`, the sequence is:
```
1² + 9² = 82
8² + 2² = 68
6² + 8² = 100
1² + 0² + 0² = 1
```
Thus, **19 is a happy number**.

**Tasks**:
1. Write a function `is_happy(n)` that returns `True` if `n` is happy.
2. Write a function `happy(n)` that returns a list of all happy numbers **less than `n`**.

**Example Output**:

````python
happy(100) = [1, 7, 10, 13, 19, 23, 28, 31, 32, 44, 49, 68, 70, 79, 82, 86, 91, 94, 97]
````

## **Exercise: Polynomial Derivative**

A **polynomial** can be represented as a Python list of its coefficients.
For example, `[1, 5, -4]` represents the polynomial $1 + 5x - 4x^2$.

**Task**:
Write a function `diff(P, n)` that returns the **n-th derivative** of the polynomial `P`.
**Do not use external packages**.

**Examples**:

````python
diff([3, 2, 1, 5, 7], 2) = [2, 30, 84]
diff([-6, 5, -3, -4, 3, -4], 3) = [-24, 72, -240]
````
