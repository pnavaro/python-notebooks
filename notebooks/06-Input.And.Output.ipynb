## **String Representations: `str()` vs. `repr()`**

- **`str()`**: Returns a **human-readable** representation of a value.
- **`repr()`**: Returns a **machine-readable** representation that can be used to recreate the object using `eval()`.
- For objects without a specific human-readable format, `str()` returns the same value as `repr()`.

**Examples:**

````python
s = 'Hello, world.'
print(str(s))   # Output: Hello, world.
print(repr(s))  # Output: 'Hello, world.'

l = list(range(4))
print(str(l))   # Output: [0, 1, 2, 3]
print(repr(l))  # Output: [0, 1, 2, 3]
````

**Note:** `repr()` of a string adds **quotes and escape characters** for special symbols (e.g., `\n`).

````python
hello = 'hello, world\n'
print(repr(hello))  # Output: 'hello, world\n'
````

`repr()` can also represent **any Python object**:

````python
x, y = 10 * 3.25, 200 * 200
print(repr((x, y, ('spam', 'eggs'))))  # Output: (32.5, 40000, ('spam', 'eggs'))
````

## **Formatting Output**

### **Using `rjust()` for Alignment**

````python
n = 7
for x in range(1, n):
    for i in range(n):
        print(repr(x**i).rjust(i + 2), end=' ')
    print()
````

### **Using C-Style Formatting**

````python
for x in range(1, n):
    for i in range(n):
        print("%07d" % x**i, end=' ')  # Pads with zeros to 7 digits
    print()
````

## **Using `str.format()` for String Formatting**

The `str.format()` method provides a powerful way to format strings. It allows for **placeholders** and **advanced formatting options**.

**Example:**

````python
x = 10 * 3.25
y = 200 * 200
s = 'The value of x is {}, and y is {}...'.format(x, repr(y))
print(s)  # Output: The value of x is 32.5, and y is 40000...
````

## **File Handling and Text Processing**

### **Reading and Processing Text Files**

**Example: Counting Words in a File**

````python
def words(filename):
    """Parse a file and return a sorted list of words."""
    with open(filename, 'r') as file:
        return sorted(file.read().split())

word_list = words('sample.txt')
print(word_list[:5])  # Example output: ['adipisci', 'adipisci', 'aliquam', ...]
````

### **Counting Word Occurrences**

````python
def reduce(word_list):
    """Count the number of occurrences of each word in a list and return a dictionary."""
    word_count = {}
    for word in word_list:
        word_count[word] = word_count.get(word, 0) + 1
    return word_count

word_counts = reduce(words('sample.txt'))
print(word_counts)
# Example output: {'neque': 80, 'ut': 80, 'est': 76, 'amet': 74, ...}
````

## **Saving Structured Data with JSON**

**JSON (JavaScript Object Notation)** is a popular format for **data interchange** between applications, even those written in different languages.

### **Serializing Data to JSON**

````python
import json

# Convert a Python object to a JSON string
json_string = json.dumps([1, 'simple', 'list'])
print(json_string)  # Output: "[1, "simple", "list"]"

# Save a dictionary to a JSON file
x = dict(name="Pierre Navaro", organization="CNRS", position="IR")
with open('workfile.json', 'w') as f:
    json.dump(x, f)
````

### **Deserializing JSON Data**

````python
# Load JSON data from a file
with open('workfile.json', 'r') as f:
    loaded_data = json.load(f)
print(loaded_data)  # Output: {'name': 'Pierre Navaro', 'organization': 'CNRS', 'position': 'IR'}
````
