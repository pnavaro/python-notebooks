## **Introduction to Classes**

Classes in Python provide a way to **bundle data and functionality together**. They allow you to:
- Create **new types** of objects.
- Define **instances** of those types.
- Attach **attributes** and **methods** to instances.
- Use **inheritance** to create hierarchical relationships between classes.

---

## **Using Classes to Store Data**

An empty class can be used to **bundle data items** together, similar to a struct in other languages.

**Example: Simple Class for Storing Data**

```python
class Car:
    pass

mycar = Car()  # Create an empty Car instance

# Add attributes to the instance
mycar.brand = 'Peugeot'
mycar.model = 308
mycar.year = 2015

# View the instance's attributes
print(mycar.__dict__)  # Output: {'brand': 'Peugeot', 'model': 308, 'year': 2015}
```

---
## **Using `namedtuple` for Immutable Data**

The `namedtuple` from the `collections` module provides a **lightweight, immutable** way to bundle data.

**Example:**

```python
from collections import namedtuple

# Define a namedtuple type
Car = namedtuple('Car', ['brand', 'model', 'year'])

# Create an instance
mycar = Car('Peugeot', 308, 2015)
print(mycar)  # Output: Car(brand='Peugeot', model=308, year=2015)

# Access attributes
print(mycar.year)  # Output: 2015

# namedtuples are immutable
import sys
try:
    mycar.model = 3008  # Attempt to modify
except:
    print(sys.exc_info()[0])  # Output: <class 'AttributeError'>
```

## **Defining Classes with Methods**

A class can include **methods** to define behavior for its instances.

**Example: Class with Constructor and Methods**

```python
class Car:
    """A simple class representing a Car with brand, model, and year."""

    def __init__(self, brand, model, year):
        """Constructor to initialize the Car instance."""
        self.brand = brand
        self.model = model
        self.year = year

    def __repr__(self):
        """String representation of the Car instance."""
        return f"Car(brand={self.brand!r}, model={self.model!r}, year={self.year!r})"

# Create an instance
mycar = Car('Peugeot', 308, 2015)
print(mycar)  # Output: Car(brand='Peugeot', model=308, year=2015)
```

## **Class Attributes and Static Methods**

You can define **class attributes** and **static methods** that are shared across all instances of a class.

**Example: Class with Class Attributes**

```python
class MyComplex:
    """A class representing a complex number with a precision attribute."""

    precision = 3  # Class attribute

    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __repr__(self):
        return f"{self.real:.{self.precision}f} + {self.imag:.{self.precision}f}j"

# Create instances
u = MyComplex(1.23456, 7.89012)
v = MyComplex(3.45678, 9.01234)

# Modify class attribute
MyComplex.precision = 1
print(u)  # Output: 1.2 + 7.9j
print(v)  # Output: 3.5 + 9.0j

# Reset class attribute
MyComplex.precision = 5
print(u)  # Output: 1.23456 + 7.89012j
print(v)  # Output: 3.45678 + 9.01234j
```

## **Example: Rational Number Class**

Here’s a class to represent **rational numbers** with arithmetic operations.

```python
class Rational:
    """A class representing a rational number."""

    def __init__(self, n, d):
        """Initialize the rational number and simplify it."""
        assert isinstance(n, int) and isinstance(d, int), "Numerator and denominator must be integers."

        def gcd(x, y):
            """Compute the greatest common divisor of x and y."""
            if x == 0:
                return y
            elif x < 0:
                return gcd(-x, y)
            elif y < 0:
                return -gcd(x, -y)
            else:
                return gcd(y % x, x)

        g = gcd(n, d)
        self.numer = n // g
        self.denom = d // g

    def __add__(self, other):
        """Add two rational numbers."""
        return Rational(
            self.numer * other.denom + other.numer * self.denom,
            self.denom * other.denom
        )

    def __sub__(self, other):
        """Subtract two rational numbers."""
        return Rational(
            self.numer * other.denom - other.numer * self.denom,
            self.denom * other.denom
        )

    def __mul__(self, other):
        """Multiply two rational numbers."""
        return Rational(
            self.numer * other.numer,
            self.denom * other.denom
        )

    def __truediv__(self, other):
        """Divide two rational numbers."""
        return Rational(
            self.numer * other.denom,
            self.denom * other.numer
        )

    def __repr__(self):
        """String representation of the rational number."""
        return f"{self.numer:d}/{self.denom:d}"

# Example usage
r1 = Rational(2, 3)
r2 = Rational(3, 4)
print(r1 + r2, r1 - r2, r1 * r2, r1 / r2)  # Output: 17/12, -1/12, 1/2, 8/9
```
