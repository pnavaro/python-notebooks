## **Overview of Matplotlib**

[Matplotlib](https://matplotlib.org/) is a **comprehensive 2D plotting library** for Python. It enables the creation of **publication-quality figures** in various formats and interactive environments. Matplotlib is designed to:
- Make **easy things easy** and **hard things possible**.
- Generate **plots, histograms, bar charts, scatter plots, power spectra, error charts**, and more with just a few lines of code.

**Key Features:**
- **MATLAB-like interface** via the `pyplot` module.
- **Highly customizable** plots.
- **Integration** with Jupyter Notebook for inline or interactive plots.
- **Extensive gallery** of examples: [Matplotlib Gallery](https://matplotlib.org/gallery.html).

**Note:** Some examples in this notebook use NumPy commands, which will be explained in more detail later.

---

## **Setting Up Matplotlib**

To use Matplotlib effectively in Jupyter Notebook:
- Use `%matplotlib inline` for **static inline plots**.
- Use `%matplotlib notebook` for **interactive plots**.
- Configure the display for **high-resolution** outputs (e.g., Retina screens).

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Configure for high-resolution displays
%config InlineBackend.figure_format = "retina"

# Set default figure size
plt.rcParams['figure.figsize'] = (10.0, 6.0)  # Wider figures for better visibility
```

## **Line Plots**

### **Basic Line Plot**
`np.linspace(start, stop, num)` generates `num` **evenly spaced values** between `start` and `stop`.

**Example: Sinc Function**

```python
x = np.linspace(-5 * np.pi, 5 * np.pi, 100)
plt.plot(x, np.sin(x) / x)
plt.title("Sinc Function: sin(x)/x")
plt.xlabel("x")
plt.ylabel("sin(x)/x")
plt.grid(True)
plt.show()
```

**Output:** A plot of the sinc function over the range `[-5π, 5π]`.

---
### **Multiple Line Plots**
You can plot **multiple functions** on the same graph.

**Example: Sinc and Modified Sinc**

```python
plt.plot(x, np.sin(x) / x, label="sin(x)/x")
plt.plot(x, np.sin(2 * x) / x, label="sin(2x)/x")
plt.legend()
plt.title("Comparison of Sinc Functions")
plt.xlabel("x")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

**Output:** A plot comparing `sin(x)/x` and `sin(2x)/x`.

## **Animations with Matplotlib**

Matplotlib supports **animations** for dynamic visualizations. Here’s an example of a simple animation:

### **Example: Wave Animation**

```python
from matplotlib import animation
from IPython.display import HTML

def basic_animation(dt=0.1):
    """Create a basic animation of a wave."""
    fig, ax = plt.subplots()
    x = np.linspace(0, 2 * np.pi, 100)
    line, = ax.plot(x, np.sin(x))

    def animate(i):
        line.set_ydata(np.sin(x + i * dt))  # Update the data
        return line,

    return animation.FuncAnimation(fig, animate, frames=100, interval=50)

# Display the animation
HTML(basic_animation(dt=0.3).to_jshtml())
```

**Output:** An animated wave plot.

## **Advanced Plotting Features**

Matplotlib supports a wide range of **customizations** and **advanced plotting features**, such as:
- **Annotations** and **text**.
- **Quiver plots** for vector fields.
- **Custom styling** (colors, line styles, markers).
- **Subplots** for multiple plots in a single figure.

**Example: Customized Plot with Annotations**

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label="sin(x)")
ax.set_title("Customized Sine Wave")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.grid(True)
ax.legend()
ax.annotate('Peak', xy=(np.pi/2, 1), xytext=(np.pi/2 + 1, 1.5),
            arrowprops=dict(facecolor='black', shrink=0.05))
plt.show()
```
