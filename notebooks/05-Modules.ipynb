## **Introduction to Modules**

As your Python programs grow, you may want to **split them into multiple files** for easier maintenance. Python supports this by allowing you to define functions, classes, and variables in a file and reuse them in other scripts or interactive sessions. Such a file is called a **module**.

---

## **Creating a Module**

Run the following cell to create a file named `fibo.py` with two functions to compute the Fibonacci series:

````python
%%file fibo.py
"""Simple module with two functions to compute Fibonacci series."""

def fib1(n):
    """Print Fibonacci series up to n."""
    a, b = 0, 1
    while b < n:
        print(b, end=', ')
        a, b = b, a + b

def fib2(n):
    """Return Fibonacci series up to n as a list."""
    result = []
    a, b = 0, 1
    while b < n:
        result.append(b)
        a, b = b, a + b
    return result

if __name__ == "__main__":
    import sys
    fib1(int(sys.argv[1]))
````

---
## **Importing and Using Modules**

You can use the functions in `fibo.py` by importing the module (without the `.py` extension).

````python
import fibo
print(fibo.__name__)  # Output: fibo
print(fibo.__file__)  # Output: Path to fibo.py
fibo.fib1(1000)       # Prints Fibonacci series up to 1000
````

To run the module as a script:

````python
%run fibo.py 1000
````

To view the module's documentation:

````python
help(fibo)
````

---
## **Executing Modules as Scripts**

When you run a Python module from the command line:

````python
%%bash
python fibo.py <arguments>
````

The code in the module is executed, and the `__name__` variable is set to `"__main__"`. This allows you to write code that behaves differently when run as a script versus when imported as a module.

## **Module Search Path**

When you import a module, Python searches for it in the following order:
1. The current directory.
2. Directories listed in the `PYTHONPATH` environment variable.
3. The installation-dependent default directory (e.g., `site-packages`).

You can also add directories to the search path dynamically:

````python
import sys
sys.path.append("/Users/navaro/python-notebooks/")
print(sys.path)
````

When importing a module named `foo`, Python searches for the following files in the specified order:
- `foo.dll`, `foo.dylib`, or `foo.so` (compiled extensions)
- `foo.py` (source file)
- `foo.pyc` (compiled file)
- `foo/__init__.py` (package directory)

## **Best Practices for Imports**

- **Avoid `from module import *`**: This can lead to **namespace pollution** and make it unclear which functions or variables are being used.

**Example of Namespace Pollution:**

````python
sum(range(5), -1)  # Uses Python's built-in sum function

from numpy import *
sum(range(5), -1)  # Uses NumPy's sum function, which behaves differently
````

To resolve this, you can delete the imported function:

````python
del sum  # Delete the imported sum function from NumPy
help(sum)  # Now refers to Python's built-in sum
````

**Recommended Approach:**

````python
import numpy as np
help(np.sum)  # Use the module prefix to avoid conflicts
````
