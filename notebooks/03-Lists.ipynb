# Python Lists and Tuples

---

## **Introduction to Lists and Tuples**

- **Lists** are the most versatile Python data type for grouping values.
  - Written as a list of **comma-separated values** between **square brackets** (`[]`).
  - Can contain items of **different types**.
  - Support **indexing, slicing, and concatenation** (like strings).
  - **Mutable**: Can be modified after creation.

- **Tuples** are similar to lists but are **immutable** (read-only).
  - Written between **parentheses** (`()`).

---

## **Indexing and Slicing**

```python
squares = [1, 4, 9, 16, 25]
print(squares)  # Output: [1, 4, 9, 16, 25]

print(squares[0])   # Indexing returns the item → 1
print(squares[-1])  # Last item → 25
print(squares[-3:]) # Slicing returns a new list → [9, 16, 25]
```

---
## **Modifying Lists**

Unlike strings, **lists are mutable**, meaning you can modify them after creation.

```python
squares += [36, 49, 64, 81, 100]  # Concatenate lists
print(squares)  # Output: [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]

cubes = [1, 8, 27, 65, 125]  # There's an error here!
cubes[3] = 64  # Replace the wrong value (cube of 4 is 64, not 65)
print(cubes)   # Output: [1, 8, 27, 64, 125]

cubes.append(216)  # Add the cube of 6
print(cubes)   # Output: [1, 8, 27, 64, 125, 216]

cubes.remove(1)  # Remove the first occurrence of 1
print(cubes)   # Output: [8, 27, 64, 125, 216]
```

---
## **List Assignment and References**

- You can **change the size** of a list or **clear it entirely**.
- The `len()` function returns the **size of the list**.
- **Assignment behavior**:
  - Assigning a list to another variable creates a **reference** (both variables point to the same list in memory).
  - To create a **new independent list**, use slicing (`[:]`).

```python
a = [1, 2, 3, 4, 5]
b = a  # b is a reference to a
b[2] = 10
print("b =", b)  # Output: b = [1, 2, 10, 4, 5]
print("a =", a)  # Output: a = [1, 2, 10, 4, 5] (a is also modified)

b = a[:]  # Create a new list by slicing
b[2] = 20
print("b =", b)  # Output: b = [1, 2, 20, 4, 5]
print("a =", a)  # Output: a = [1, 2, 10, 4, 5] (a remains unchanged)
```

## **Useful List Methods**

```python
a = list("Python-2020")
print(a)  # Output: ['P', 'y', 't', 'h', 'o', 'n', '-', '2', '0', '2', '0']

a.sort()  # Sort the list in place
print(a)  # Output: ['-', '0', '0', '2', '2', 'P', 'h', 'n', 'o', 't', 'y']

a.reverse()  # Reverse the list in place
print(a)  # Output: ['y', 't', 'o', 'n', 'h', 'P', '2', '2', '0', '0', '-']

a.pop()  # Remove and return the last item
print(a)  # Output: ['y', 't', 'o', 'n', 'h', 'P', '2', '2', '0', '0']
```

## **Introduction to Dictionaries**

Dictionaries are **collections of key-value pairs**, indexed by **keys** (often strings).

```python
person = dict(firstname="John", lastname="Smith", email="john.doe@domain.fr")
person['size'] = 1.80  # Add a new key-value pair
person['weight'] = 70   # Add another key-value pair
print(person)  # Output: {'firstname': 'John', 'lastname': 'Smith', 'email': 'john.doe@domain.fr', 'size': 1.8, 'weight': 70}

print(person.keys())   # Output: dict_keys(['firstname', 'lastname', 'email', 'size', 'weight'])
print(person.items())  # Output: dict_items([('firstname', 'John'), ('lastname', 'Smith'), ...])
```

## **Exercises**

1. Split the string `"python ENSAI 2020"` into the list `["python", "ENSAI", 2020]`.
2. Insert `"september"` and the value `7` before `2020` in the resulting list.
3. Capitalize the first item to `"Python"`.
4. Create a dictionary with the following keys: `meeting`, `month`, `day`, `year`.
5. Print out the items of the dictionary.
6. Append the key `"place"` to the dictionary and set its value to `"ENSAI"`.

**Solution:**

```python
# Step 1: Split the string
result = ["python", "ENSAI", 2020]

# Step 2: Insert "september" and 7 before 2020
result.insert(-1, 7)
result.insert(-1, "september")

# Step 3: Capitalize the first item
result[0] = result[0].capitalize()

# Step 4: Create a dictionary
exercise_dict = {
    "meeting": result[0],
    "month": result[2],
    "day": result[1],
    "year": result[3]
}

# Step 5: Print the items
print(exercise_dict.items())

# Step 6: Append the key "place"
exercise_dict["place"] = "ENSAI"
print(exercise_dict)
```
