# Errors and Exceptions in Python

---

## **Types of Errors**

There are **two main types of errors** in Python:

1. **Syntax Errors (Parsing Errors)**
   - Occur when the code violates Python's syntax rules.
   - Detected by the interpreter during parsing.

2. **Exceptions**
   - Occur when the code is syntactically correct but encounters an unexpected condition during execution.
   - **Not fatal**: Can be handled using `try-except` blocks.

For more details, refer to the [Python Exceptions Documentation](https://docs.python.org/3/library/exceptions.html).

---

## **Common Exceptions**

### **Division by Zero**

```python
import sys
try:
    10 * (1 / 0)
except:
    print(sys.exc_info()[0])  # Output: <class 'ZeroDivisionError'>
```

### **Undefined Variable**

```python
try:
    4 + spam * 3
except:
    print(sys.exc_info()[0])  # Output: <class 'NameError'>
```

### **Type Error**

```python
try:
    '2' + 2
except:
    print(sys.exc_info()[0])  # Output: <class 'TypeError'>
```

---

## **Handling Exceptions**

### **Basic `try-except` Block**
Use `try-except` to handle exceptions gracefully.

**Example: Converting User Input**

```python
for s in ("0.1", "foo", "1000"):
    try:
        x = int(s)
        print(f'x = {x}')
        break
    except ValueError:
        print("Oops! That was no valid number. Try again...")
```

### **Multiple `except` Clauses**
A `try` statement can have **multiple `except` clauses** to handle different exceptions.

### **The `else` Clause**
The `else` clause runs if **no exception occurs** in the `try` block.

---

## **Example: Processing a File**

Here’s a function that reads a file, converts its first line to an integer, and checks if it’s positive.

```python
import sys

def process_file(file):
    """Read the first line of a file, convert it to an integer, and check if it is positive."""
    try:
        with open(file, 'r') as f:
            i = int(f.readline())
            assert i > 0, "Integer must be positive"
            print(f"Integer: {i}")
    except ValueError:
        print("Error: First line is not a valid integer.")
    except AssertionError as e:
        print(f"Error: {e}")
    except FileNotFoundError:
        print("Error: File not found.")
    except PermissionError:
        print("Error: Permission denied.")
    except Exception as e:
        print(f"Unexpected error: {e}")
```

### **Testing the Function**

#### **Case 1: Valid File with Valid Integer**

```python
# Create a file with a valid integer
with open('workfile.txt', 'w') as f:
    f.write("42")

process_file('workfile.txt')  # Output: Integer: 42
```

#### **Case 2: File with Invalid Integer**

```python
# Insert an invalid integer
!echo "-1" > workfile.txt
process_file('workfile.txt')  # Output: Error: Integer must be positive
```

#### **Case 3: File with Non-Integer Content**

```python
# Insert non-integer content
!echo "foo" > workfile.txt
process_file('workfile.txt')  # Output: Error: First line is not a valid integer.
```

#### **Case 4: File Does Not Exist**

```python
# Delete the file
!rm -f workfile.txt
process_file('workfile.txt')  # Output: Error: File not found.
```

#### **Case 5: Permission Denied**

```python
# Change file permissions to deny read access
!chmod u-r workfile.txt
process_file('workfile.txt')  # Output: Error: Permission denied.
```

## **Raising Exceptions**

You can **force an exception** to occur using the `raise` statement.

**Example:**

```python
try:
    raise NameError('HiThere')
except:
    print(sys.exc_info()[0])  # Output: <class 'NameError'>
```

---
## **Clean-up Actions with `finally`**

The `finally` clause is **always executed** before leaving the `try` statement, regardless of whether an exception occurred.

**Example:**

```python
try:
    raise KeyboardInterrupt
except:
    print(sys.exc_info()[0])  # Output: <class 'KeyboardInterrupt'>
finally:
    print('Goodbye, world!')  # Output: Goodbye, world!
```

## **Exercise: Word Count with Exception Handling**

**Task:**
Improve the `reduce` function to handle the results of the `words` function by using the `KeyError` exception to fill in the dictionary.

**Example:**

```python
def reduce(word_list):
    """Count the number of occurrences of each word in a list and return a dictionary."""
    word_count = {}
    for word in word_list:
        try:
            word_count[word] += 1
        except KeyError:
            word_count[word] = 1
    return word_count
```
