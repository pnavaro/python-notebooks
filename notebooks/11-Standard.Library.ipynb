## **Operating System Interface**

The `os` module provides a **portable way** of using operating system-dependent functionality, such as reading or writing files, and running system commands.

### **Common Operations**

**Get the current working directory:**

```python
import os
print(os.getcwd())  # Output: Current working directory (e.g., '/home/user')
```

**Set Environment Variables (e.g., for macOS):**

```python
import sys
if sys.platform == "darwin":
    os.environ['CC'] = 'gcc-10'  # Change the default C compiler to gcc-10 on macOS
```

**Run System Commands:**

```python
os.system('mkdir today')  # Run the command `mkdir today` in the system shell
```

**Change Directory and Create Files:**

```python
os.chdir('today')         # Change current working directory to 'today'
os.system('touch data.db')  # Create an empty file named 'data.db'
```

**Copy, Remove, and Move Files:**

```python
import shutil
shutil.copyfile('data.db', 'archive.db')  # Copy 'data.db' to 'archive.db'

if os.path.exists('backup.db'):  # Check if 'backup.db' exists
    os.remove('backup.db')       # Remove 'backup.db'

shutil.move('archive.db', 'backup.db')  # Move 'archive.db' to 'backup.db'
os.chdir('..')  # Move back to the parent directory
```

## **Recursive File Operations**

**Example: Recursively Replace a String in Files**

```python
def recursive_replace(root, pattern, replace):
    """
    Replace a string in all files within a directory.

    Args:
        root (str): Directory to search in.
        pattern (str): String to search for.
        replace (str): String to replace with.
    """
    for directory, subdirs, filenames in os.walk(root):
        for filename in filenames:
            path = os.path.join(directory, filename)
            with open(path, 'r') as file:
                text = file.read()
            if pattern in text:
                print(f'Occurrence in: {filename}')
                text = text.replace(pattern, replace)
                with open(path, 'w') as file:
                    file.write(text)
```

## **Performance Measurement**

The `timeit` module provides a simple way to **measure the execution time** of small bits of Python code.

**Example: Compare Two Approaches for Swapping Variables**

```python
from timeit import Timer

# Method 1: Using a temporary variable
Timer('t=a; a=b; b=t', 'a=1; b=2').timeit()

# Method 2: Pythonic way
Timer('a,b = b,a', 'a=1; b=2').timeit()

# Using %%timeit magic in Jupyter
%%timeit a=1; b=2
a, b = b, a
```

## **Profiling Code**

The [`profile`](https://docs.python.org/3/library/profile.html#module-profile) and [`pstats`](https://docs.python.org/3/library/profile.html#module-pstats) modules provide tools for **identifying time-critical sections** in larger blocks of code.

**Example:**

```python
import cProfile
import pstats

# Profile a function
cProfile.run('my_function()', 'output.prof')

# Analyze the profiling results
stats = pstats.Stats('output.prof')
stats.sort_stats('cumulative').print_stats(10)  # Print the top 10 time-consuming calls
```

## **Quality Control with `doctest`**

The `doctest` module provides a tool for **scanning a module and validating tests embedded in docstrings**. This improves documentation and ensures code remains true to the documentation.

**Example:**

```python
def average(values):
    """Computes the arithmetic mean of a list of numbers.

    >>> print(average([20, 30, 70]))
    40.0
    """
    return sum(values) / len(values)

import doctest
doctest.testmod()  # Automatically validate the embedded tests
```
