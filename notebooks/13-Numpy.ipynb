## **What Does NumPy Provide?**

NumPy (Numerical Python) is a **fundamental package** for scientific computing in Python. It provides:

- **`ndarray`**: A powerful **n-dimensional array object** for efficient data storage and manipulation.
- **Derived objects**: Such as masked arrays and matrices.
- **`ufunc`**: Fast **universal functions** for array operations.
- **MATLAB-like capabilities**: Enables MATLAB-style operations within Python.
- **Developed by [Travis Oliphant](https://www.continuum.io/people/travis-oliphant)**.
- **NumPy 1.0** was released in **October 2006**.

For more details, refer to the **[NumPy Documentation](https://docs.scipy.org/doc/numpy/)**.

---

## **Routines for Fast Array Operations**

NumPy provides **optimized routines** for a wide range of operations on arrays, including:
- **Shape manipulation** (reshaping, stacking, splitting).
- **Sorting** (quick sort, merge sort, partial sort).
- **Input/Output (I/O)** (saving/loading arrays to/from files).
- **Fast Fourier Transform (FFT)**.
- **Basic linear algebra** (matrix multiplication, determinants, eigenvalues).
- **Basic statistical operations** (mean, median, variance, standard deviation).
- **Random simulation** (random number generation, shuffling).
- **Statistics** (histograms, correlations, regressions).
- And **much more**...

## **Why Use NumPy Arrays?**

### **Limitations of Python Lists**
- **Slow processing**: Python lists are **slow** for numerical operations.
- **Memory inefficiency**: Lists use **more memory** than NumPy arrays.
- **Complexity**: For **multi-dimensional data** (e.g., matrices, tensors), you need nested lists, which become **messy and hard to manage**.

### **Performance Comparison: Lists vs. NumPy Arrays**

```python
from random import random
from operator import truediv
from timeit import timeit

# Using Python lists
def python_sum(n):
    a = [random() for _ in range(n)]
    b = [random() for _ in range(n)]
    return sum(x / y for x, y in zip(a, b))

# Using NumPy arrays
def numpy_sum(n):
    a = np.random.random(n)
    b = np.random.random(n)
    return np.sum(a / b)

n = 10**6
print("Python list time:", timeit(lambda: python_sum(n), number=10))
print("NumPy array time:", timeit(lambda: numpy_sum(n), number=10))
```

**Output:**
- Python list time: ~5.0 seconds
- NumPy array time: ~0.1 seconds

## **NumPy Array Programming**

### **Key Principles**
- **Avoid loops**: NumPy array operations are **fast**, while Python loops are **slow**.
- **Vectorized operations**: Perform operations on **entire arrays** rather than element-by-element.
- **Change of habits**: Requires a shift from traditional Python programming to **array-based programming**.
- **Experience**: Mastery of NumPy requires **practice** and familiarity with its functions.

**Example: Vectorized Operations**

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Element-wise addition
print(a + b)  # Output: [5 7 9]

# Element-wise multiplication
print(a * b)  # Output: [ 4 10 18]

# Dot product
print(np.dot(a, b))  # Output: 32
```

## **Fast Evaluation of Array Expressions with `numexpr`**

The [`numexpr`](https://github.com/pydata/numexpr) package provides **fast evaluation of array expressions** by using a **vector-based virtual machine**. It also **caches expressions**, making reuse faster.

**Example: Benchmarking NumPy vs. NumExpr**

```python
import numexpr as ne
import numpy as np

nrange = (2 ** np.arange(6, 24)).astype(int)
t_numpy = []
t_numexpr = []

for n in nrange:
    a = np.random.random(n)
    b = np.arange(n, dtype=np.double)
    c = np.random.random(n)

    # NumPy evaluation
    c_numpy = a ** 2 + b ** 2 + 2 * a * b * c

    # NumExpr evaluation
    c_numexpr = ne.evaluate("a ** 2 + b ** 2 + 2 * a * b * c", optimization='aggressive')

    t1 = %timeit -oq -n 10 a ** 2 + b ** 2 + 2 * a * b * c
    t2 = %timeit -oq -n 10 ne.re_evaluate()

    t_numpy.append(t1.best)
    t_numexpr.append(t2.best)

# Plot the results
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

plt.loglog(nrange, t_numpy, label='NumPy')
plt.loglog(nrange, t_numexpr, label='NumExpr')
plt.legend(loc='lower right')
plt.xlabel('Vector Size')
plt.ylabel('Execution Time (s)')
plt.title('NumPy vs. NumExpr Performance')
plt.show()
```

**Output:** A log-log plot comparing the execution time of NumPy and NumExpr for different vector sizes.
