## **Symbolic Mathematics with SymPy**

SymPy is a **Python library for symbolic mathematics**. It provides tools for:
- **Symbolic derivatives** and integrals.
- **Algebraic manipulations** (simplification, expansion, factorization).
- **Equation solving** (algebraic, differential, and transcendental).
- **Translating mathematics into low-level code** (e.g., C, Fortran).
- **Handling very large expressions** efficiently.
- **Optimizing code using mathematical transformations**.

## **Basic Symbolic Operations**

### **Defining Symbols**

```python
x = Symbol('x')
expression = (sym.pi + x)**2
print(expression)  # Output: (π + x)²
```

### **Multiple Symbols**

```python
alpha1, omega_2 = symbols('alpha1 omega_2')
print(alpha1, omega_2)  # Output: alpha1, omega_2
```

### **Symbolic Expressions with Constraints**

```python
mu, sigma = sym.symbols('mu sigma', positive=True)
normal_distribution = 1/sym.sqrt(2 * sym.pi * sigma**2) * sym.exp(-(x - mu)**2 / (2 * sigma**2))
print(normal_distribution)
```

## **Rational Numbers in SymPy**

In Python, dividing two integers (e.g., `1/2`) results in a **float** (`0.5`). To work with **rational numbers**, use `sym.S(1)/2` or `sym.Rational(1, 2)`.

**Example:**

```python
print(x + sym.S(1)/2)  # Output: x + 1/2
print(sym.Rational(1, 4))  # Output: 1/4
```

---
## **Symbolic Operations**

### **Exponentiation vs. XOR**
- **`x ** y`**: Symbolic exponentiation (e.g., `x` raised to the power of `y`).
- **`x ^ y`**: Bitwise XOR (logical operation, not mathematical exponentiation).

**Example:**

```python
y = Symbol('y')
print(x ** y)  # Output: x^y (symbolic exponentiation)
print(x ^ y)   # Output: x XOR y (bitwise XOR)
```

## **Exercise: Bézier Curve**

A **Bézier curve** is a parametric curve used in computer graphics and animation. It is defined using **control points** and **Bernstein polynomials**.

### **Bernstein Basis Polynomials**
The Bernstein basis polynomials of degree `n` are defined as:
$$
b_{i,n}(t) = \binom{n}{i} t^i (1 - t)^{n - i}, \quad i = 0, \ldots, n
$$
where $\binom{n}{i}$ is the **binomial coefficient**.

### **Bézier Curve Definition**
The Bézier curve is a linear combination of Bernstein basis polynomials:
$$
B_n(t) = \sum_{i=0}^{n} \beta_i b_{i,n}(t)
$$

**Task:**
1. Write a function `bpoly(t, n, i)` that returns the Bernstein basis polynomial $b_{i,n}(t)$ using `sympy.binomial`.
2. Compute the Bernstein polynomial representing the Bézier curve between points $p_0$, $p_1$, and $p_2$ with $\beta_i = 1$.
3. Plot the Bézier curve for three positions of $p_1$: `(0, 0)`, `(0.5, 0.5)`, and `(1, 1)`.

**Given Points:**
$$
p_0 = (1, 0); \qquad p_1 = (x, y); \qquad p_2 = (0, 1)
$$

**Solution Outline:**

```python
from sympy import binomial

def bpoly(t, n, i):
    """Return the Bernstein basis polynomial b_{i,n}(t)."""
    return binomial(n, i) * t**i * (1 - t)**(n - i)

# Define symbols
t = Symbol('t')
p0 = (1, 0)
p1 = (x, y)
p2 = (0, 1)

# Compute Bernstein polynomials for n=2
b0 = bpoly(t, 2, 0)
b1 = bpoly(t, 2, 1)
b2 = bpoly(t, 2, 2)

# Bézier curve
B = b0 * p0[0] + b1 * p1[0] + b2 * p2[0], b0 * p0[1] + b1 * p1[1] + b2 * p2[1]
print(B)
```

## **Numerical Integration with SymPy**

SymPy provides tools for **numerical integration**, including **Gaussian quadrature** methods.

### **Gauss-Legendre Quadrature**

```python
from sympy.integrals.quadrature import gauss_legendre

# Compute nodes and weights for Gauss-Legendre quadrature
x, w = gauss_legendre(3, 5)  # 3 points, 5th order
print("Nodes:", x)
print("Weights:", w)
```

### **Gauss-Lobatto Quadrature**

```python
from sympy.integrals.quadrature import gauss_lobatto

# Compute nodes and weights for Gauss-Lobatto quadrature
x, w = gauss_lobatto(3, 12)  # 3 points, 12th order
print("Nodes:", x)
print("Weights:", w)
```
